```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Optional: nicer default plot style
sns.set_theme(style="whitegrid")

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
titanic = pd.read_csv(url)

print(titanic.shape)
titanic.head(10)

titanic.info()
titanic.describe()

# Group ages into bins and calculate survival rate per bin
age_bins = pd.cut(titanic['Age'], bins=range(0, 90, 5))
survival_by_age = titanic.groupby(age_bins)['Survived'].mean()

plt.figure(figsize=(10, 6))
plt.plot(survival_by_age.index.astype(str), survival_by_age.values, marker='o', color='steelblue')
plt.title("Age vs Survival")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(data=titanic, x="Age", hue="Survived", bins=30, kde=True, palette=["salmon", "seagreen"], alpha=0.6)

plt.title("Age Distribution by Survival")
plt.xlabel("Age")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
